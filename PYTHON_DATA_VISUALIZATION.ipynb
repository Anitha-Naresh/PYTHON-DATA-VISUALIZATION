```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

```python
df = sns.load_dataset("taxis")

print("Dataset loaded successfully!")
print("\nFirst 5 rows:")
display(df.head())

print("\nShape of the dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
display(df.describe(include="all"))


# ============================================================
# 4. CHECK FOR MISSING VALUES
# ============================================================

print("\nMissing Values in Each Column:")
missing_values = df.isnull().sum()
print(missing_values)

print("\nPercentage of Missing Values:")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)
```

```python
# Convert pickup to datetime
df["pickup"] = pd.to_datetime(df["pickup"], errors="coerce")

# Convert dropoff to datetime
df["dropoff"] = pd.to_datetime(df["dropoff"], errors="coerce")

# Numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

# Categorical columns
categorical_columns = df.select_dtypes(include=["object", "category"]).columns

# Impute numerical columns using median
for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())
        # Impute categorical columns using mode
for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        mode_value = df[column].mode()
        if len(mode_value) > 0:
            df[column] = df[column].fillna(mode_value[0])

            # Remove rows where critical pickup timestamp is missing
df = df.dropna(subset=["pickup"])

print("\nMissing values after handling:")
print(df.isnull().sum())

print("\nDataset shape after cleaning:")
print(df.shape)
```

```python
print("\nFirst 5 rows of cleaned dataset:")
display(df.head())

print("\nData types after cleaning:")
print(df.dtypes)
```

```python
# Sort data according to pickup time
line_df = df.sort_values("pickup").copy()

plt.figure(figsize=(14, 6))

plt.plot(
    line_df["pickup"],
    line_df["fare"],
    linewidth=1
)

plt.title("Fare Over Time", fontsize=16)
plt.xlabel("Pickup Timestamp", fontsize=12)
plt.ylabel("Fare", fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
borough_fare = (
    df.groupby("pickup_borough")["fare"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

borough_fare.plot(
    kind="bar"
)

plt.title("Total Fare by Pickup Borough", fontsize=16)
plt.xlabel("Pickup Borough", fontsize=12)
plt.ylabel("Total Fare", fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
payment_counts = df["payment"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Trips by Payment Method", fontsize=16)
plt.axis("equal")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df["distance"],
    bins=30,
    edgecolor="black"
)

plt.title("Distribution of Taxi Trip Distance", fontsize=16)
plt.xlabel("Distance", fontsize=12)
plt.ylabel("Number of Trips", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="pickup_borough",
    y="tip"
)

plt.title("Distribution of Tips by Pickup Borough", fontsize=16)
plt.xlabel("Pickup Borough", fontsize=12)
plt.ylabel("Tip Amount", fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="pickup_borough"
)

plt.title("Number of Trips by Pickup Borough", fontsize=16)
plt.xlabel("Pickup Borough", fontsize=12)
plt.ylabel("Number of Trips", fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 7))

sns.scatterplot(
    data=df,
    x="distance",
    y="fare",
    hue="pickup_borough",
    alpha=0.6
)

plt.title("Relationship Between Distance and Fare", fontsize=16)
plt.xlabel("Distance", fontsize=12)
plt.ylabel("Fare", fontsize=12)
plt.legend(title="Pickup Borough", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python

correlation_columns = [
    "distance",
    "fare",
    "tip",
    "tolls",
    "total"
]

correlation_matrix = df[correlation_columns].corr()

print("\nCorrelation Matrix:")
display(correlation_matrix)

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Taxi Data", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
pair_columns = [
    "distance",
    "fare",
    "tip",
    "total",
    "pickup_zone"
]

# Use a sample to keep the pair plot readable and fast
pair_df = df[pair_columns].dropna().copy()

if len(pair_df) > 1000:
    pair_df = pair_df.sample(1000, random_state=42)

sns.pairplot(
    pair_df,
    vars=["distance", "fare", "tip", "total"],
    hue="pickup_zone",
    diag_kind="hist"
)

plt.suptitle(
    "Pair Plot of Distance, Fare, Tip and Total by Pickup Zone",
    y=1.02,
    fontsize=16
)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="payment",
    y="fare"
)

plt.title("Distribution of Fare by Payment Method", fontsize=16)
plt.xlabel("Payment Method", fontsize=12)
plt.ylabel("Fare", fontsize=12)
plt.tight_layout()
plt.show()
```

```python

print("\n================ ADDITIONAL ANALYSIS ================")

# Average fare
print("\nAverage Fare:")
print(round(df["fare"].mean(), 2))

# Average distance
print("\nAverage Distance:")
print(round(df["distance"].mean(), 2))

# Average tip
print("\nAverage Tip:")
print(round(df["tip"].mean(), 2))

# Total fare
print("\nTotal Fare:")
print(round(df["fare"].sum(), 2))

# Total trips
print("\nTotal Number of Trips:")
print(len(df))

# Most common payment method
print("\nMost Common Payment Method:")
print(df["payment"].mode()[0])

# Most common pickup borough
print("\nMost Common Pickup Borough:")
print(df["pickup_borough"].mode()[0])

# Borough with highest total fare
print("\nPickup Borough with Highest Total Fare:")
print(borough_fare.idxmax())

# Highest total fare
print("\nHighest Total Fare by Borough:")
print(round(borough_fare.max(), 2))
```

```python
print("\n================ CORRELATION ANALYSIS ================")

print("\nCorrelation between Distance and Fare:")
distance_fare_corr = df["distance"].corr(df["fare"])
print(round(distance_fare_corr, 3))

print("\nCorrelation between Fare and Total:")
fare_total_corr = df["fare"].corr(df["total"])
print(round(fare_total_corr, 3))

print("\nCorrelation between Tip and Total:")
tip_total_corr = df["tip"].corr(df["total"])
print(round(tip_total_corr, 3))
```
